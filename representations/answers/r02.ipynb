## R02.C1: normalize_columns
Produce unit columns; reject a zero column.

Hint: The norm is taken over rows (axis=0).

```python
def normalize_columns(a):
    a=np.asarray(a,float); norms=np.linalg.norm(a,axis=0)
    if np.any(norms==0):raise ValueError('Zero direction.')
    return a/norms

assert np.allclose(np.linalg.norm(normalize_columns([[3,0],[4,2]]),axis=0),1)
print("R02.C1 reference checks passed")
```

## R02.C2: ecf_parts
Return empirical real and imaginary characteristic-function components.

Hint: Broadcast samples by frequencies and average over samples.

```python
def ecf_parts(x, t):
    phase=np.asarray(x)[:,None]*np.asarray(t)[None,:]
    return np.cos(phase).mean(0),np.sin(phase).mean(0)

t=np.linspace(-2,2,9)
a,b=ecf_parts(np.array([-1.,1.]),t)
assert np.allclose(a,np.cos(t)) and np.allclose(b,0)
print("R02.C2 reference checks passed")
```

## R02.C3: integrate_discrepancy
Integrate our weighted squared discrepancy on a strictly increasing grid.

Hint: Target and weight both equal exp(-t*t/2).

```python
def integrate_discrepancy(real, imag, t, n):
    t=np.asarray(t,float)
    if len(t)<2 or not np.all(np.diff(t)>0):raise ValueError('Increasing grid required.')
    target=np.exp(-t*t/2)
    return float(n*np.trapezoid(((np.asarray(real)-target)**2+np.asarray(imag)**2)*target,t))

t=np.linspace(-5,5,17)
assert integrate_discrepancy(np.exp(-t*t/2),np.zeros_like(t),t,10)==0
print("R02.C3 reference checks passed")
```
