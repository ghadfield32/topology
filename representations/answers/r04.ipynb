## R04.C1: accuracy_count
Return the integer correct count and accuracy; reject empty or unequal arrays.

Hint: Do not round before returning.

```python
def accuracy_count(truth, predicted):
    a=np.asarray(truth);b=np.asarray(predicted)
    if a.shape!=b.shape or a.size==0:raise ValueError('Matching nonempty arrays required.')
    correct=int(np.sum(a==b))
    return correct,correct/a.size

assert accuracy_count([0,1,2],[0,2,2])==(2,2/3)
print("R04.C1 reference checks passed")
```

## R04.C2: select_probe
Select C using ONLY pairs (C, validation_accuracy); ties choose smallest C.

Hint: Sort by decreasing validation accuracy then increasing C.

```python
def select_probe(candidates):
    if not candidates:raise ValueError('No candidates.')
    return sorted(candidates,key=lambda pair:(-pair[1],pair[0]))[0][0]

assert select_probe([(10,.8),(.1,.8),(1,.7)])==.1
print("R04.C2 reference checks passed")
```

## R04.C3: variance_magnitude
Return average sample standard deviation across embedding dimensions.

Hint: Effective rank alone ignores overall magnitude.

```python
def variance_magnitude(z):
    z=np.asarray(z,float)
    if z.ndim!=2 or len(z)<2:raise ValueError('Need a matrix with multiple observations.')
    return float(z.std(axis=0,ddof=1).mean())

assert variance_magnitude(np.ones((5,3)))==0
assert np.isclose(variance_magnitude(np.array([[1.],[3.]])),np.sqrt(2))
print("R04.C3 reference checks passed")
```
