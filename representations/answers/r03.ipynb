## R03.C1: paired_mse
Calculate mean squared difference over matched entries.

Hint: Reject a shape mismatch before subtraction.

```python
def paired_mse(a, b):
    a=np.asarray(a,float);b=np.asarray(b,float)
    if a.shape!=b.shape:raise ValueError('Shape mismatch.')
    return float(np.mean((a-b)**2))

assert paired_mse([[1,2]],[[2,4]])==2.5
print("R03.C1 reference checks passed")
```

## R03.C2: ema_update
Perform one non-mutating EMA update with 0<=tau<=1.

Hint: This is weighted averaging, not a target-network gradient.

```python
def ema_update(old, new, tau):
    if not 0<=tau<=1:raise ValueError('Invalid tau.')
    return tau*np.asarray(old)+(1-tau)*np.asarray(new)

assert np.allclose(ema_update(np.array([1.]),np.array([3.]),.75),[1.5])
print("R03.C2 reference checks passed")
```

## R03.C3: positive_pair_indices
Return targets for view A followed by view B; reject B<2.

Hint: Advance each index by B and wrap at 2B.

```python
def positive_pair_indices(batch_size):
    if batch_size<2:raise ValueError('Need at least two source observations.')
    return (np.arange(2*batch_size)+batch_size)%(2*batch_size)

assert positive_pair_indices(3).tolist()==[3,4,5,0,1,2]
print("R03.C3 reference checks passed")
```
