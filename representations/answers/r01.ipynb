## R01.C1: center_columns
Subtract one mean per column without modifying the input.

Hint: Use axis=0 and return a new array.

```python
def center_columns(x):
    x=np.asarray(x,float)
    return x-x.mean(axis=0)

x=np.array([[1.,2.],[3.,6.]])
assert np.allclose(center_columns(x).mean(0),0)
assert np.allclose(x,[[1,2],[3,6]])
print("R01.C1 reference checks passed")
```

## R01.C2: sample_covariance
Compute sample covariance, rejecting fewer than two rows.

Hint: Center then divide the Gram matrix by n-1.

```python
def sample_covariance(x):
    x=np.asarray(x,float)
    if x.ndim!=2 or len(x)<2: raise ValueError('Need two or more rows.')
    c=x-x.mean(0)
    return c.T@c/(len(x)-1)

assert np.allclose(sample_covariance([[1,2],[3,4],[5,6]]),[[4,4],[4,4]])
print("R01.C2 reference checks passed")
```

## R01.C3: apply_fitted_whitening
Apply existing mean/transform to new rows, without fitting anything.

Hint: Subtract, then matrix-multiply.

```python
def apply_fitted_whitening(x, mean, transform):
    return (np.asarray(x,float)-np.asarray(mean,float))@np.asarray(transform,float)

assert np.allclose(apply_fitted_whitening([[3.,5.]],[1,1],np.eye(2)),[[2,4]])
print("R01.C3 reference checks passed")
```
