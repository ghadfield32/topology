## R05.C1: identity_overlap
Return repeated source identities across the two partitions.

Hint: A set intersection is about identity, not row index.

```python
def identity_overlap(training_ids, test_ids):
    return sorted(set(training_ids)&set(test_ids))

assert identity_overlap([1,1,2],[2,3])==[2]
print("R05.C1 reference checks passed")
```

## R05.C2: causal_mask
Return a mask for finite observations available at or before the cutoff.

Hint: Include equality; exclude unknown timestamps.

```python
def causal_mask(timestamps, cutoff):
    t=np.asarray(timestamps,float)
    return np.isfinite(t)&(t<=cutoff)

assert causal_mask([0,.1,.2,np.nan],.1).tolist()==[True,True,False,False]
print("R05.C2 reference checks passed")
```

## R05.C3: paired_differences
Return counts newly correct and newly wrong for method B versus A.

Hint: Pair predictions on the SAME observations.

```python
def paired_differences(correct_a, correct_b):
    a=np.asarray(correct_a,bool);b=np.asarray(correct_b,bool)
    if a.shape!=b.shape:raise ValueError('Matching observations required.')
    return int((~a&b).sum()),int((a&~b).sum())

assert paired_differences([1,0,0,1],[1,1,0,0])==(1,1)
print("R05.C3 reference checks passed")
```
