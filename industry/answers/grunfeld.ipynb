## grunfeld.C1 — Lag within an entity

Sort a table by firm and year and add lag_invest from the previous row within the same firm. Keep first records missing.

Predict the check output before writing your function.

```python
def lag_investment(table):
    result=table.sort_values(['firm','year']).copy()
    result['lag_invest']=result.groupby('firm')['invest'].shift(1)
    return result
```

```python
t=pd.DataFrame({'firm':['A','B','A','B'],'year':[1,1,2,2],'invest':[10,90,20,80]})
r=lag_investment(t)
assert r.loc[2,'lag_invest']==10 and r.loc[3,'lag_invest']==90
assert r.loc[[0,1],'lag_invest'].isna().all()
print("grunfeld.C1 checks passed")
```

**Why it works:** Group boundaries are information boundaries. A global shift would mix different companies.

## grunfeld.C2 — Enforce a cutoff

Return indices with timestamp strictly earlier than the supplied decision cutoff.

Predict the check output before writing your function.

```python
def before_cutoff(times,cutoff):
    return np.flatnonzero(np.asarray(times)<cutoff)
```

```python
np.testing.assert_array_equal(before_cutoff([1947,1948,1948,1950],1948),[0])
assert len(before_cutoff([3],3))==0
print("grunfeld.C2 checks passed")
```

**Why it works:** Strictly earlier differs from earlier-or-equal. Decide how simultaneous measurements become available.

## grunfeld.C3 — Find shared entities

Return the set of entities appearing in both partitions.

Predict the check output before writing your function.

```python
def shared_entities(first,second):
    return set(first)&set(second)
```

```python
assert shared_entities(['A','B','A'],['C','B'])=={'B'}
assert shared_entities([1,2],[3])==set()
print("grunfeld.C3 checks passed")
```

**Why it works:** Entity overlap can be acceptable for future-known-firm evaluation but invalid for a new-firm claim.
