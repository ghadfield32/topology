## wdbc.C1 — Read the label map

Given loader labels 0 or 1, return integer 1 for malignant (original 0), else 0. Reject other labels.

Predict the check output before writing your function.

```python
def malignant_indicator(labels):
    a=np.asarray(labels)
    if not np.isin(a,[0,1]).all(): raise ValueError('Unknown loader label')
    return (a==0).astype(int)
```

```python
np.testing.assert_array_equal(malignant_indicator([0,1,0]),[1,0,1])
try: malignant_indicator([2])
except ValueError: pass
else: raise AssertionError('Invalid label accepted')
print("wdbc.C1 checks passed")
```

**Why it works:** Equality creates a Boolean mask; converting it to integer gives 1/0. The explicit map is necessary because numeric label order is not universal.

## wdbc.C2 — Count errors

For aligned nonempty binary vectors return (true negatives,false positives,false negatives,true positives).

Predict the check output before writing your function.

```python
def confusion_counts(actual,predicted):
    a,p=np.asarray(actual),np.asarray(predicted)
    if a.shape!=p.shape or a.ndim!=1 or not len(a) or not np.isin(a,[0,1]).all() or not np.isin(p,[0,1]).all(): raise ValueError('Binary aligned vectors required')
    return tuple(int(np.sum((a==t)&(p==q))) for t,q in [(0,0),(0,1),(1,0),(1,1)])
```

```python
assert confusion_counts([0,0,1,1],[0,1,0,1])==(1,1,1,1)
assert confusion_counts([0,1],[0,0])==(1,0,1,0)
print("wdbc.C2 checks passed")
```

**Why it works:** Count each actual/predicted pair separately; do not swap the row meaning and column meaning.

## wdbc.C3 — Calculate balanced accuracy

For aligned binary vectors with both actual classes present, average recall for class 0 and class 1.

Predict the check output before writing your function.

```python
def balanced_binary(actual,predicted):
    a,p=np.asarray(actual),np.asarray(predicted)
    if a.shape!=p.shape or a.ndim!=1 or set(a)!={0,1} or not np.isin(p,[0,1]).all(): raise ValueError('Two actual classes required')
    return float(np.mean([np.mean(p[a==c]==c) for c in (0,1)]))
```

```python
assert balanced_binary([0,0,0,1],[0,0,0,0])==.5
assert balanced_binary([0,1],[0,1])==1
print("wdbc.C3 checks passed")
```

**Why it works:** Each actual class contributes one recall regardless of how many rows belong to it. This does not certify clinical usefulness.
