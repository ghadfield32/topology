## elnino.C1 — Respect calendar order

Given YEAR and JAN..DEC columns, return a 1-D vector in year-sorted calendar order.

Predict the check output before writing your function.

```python
def calendar_vector(table):
    names=['JAN','FEB','MAR','APR','MAY','JUN','JUL','AUG','SEP','OCT','NOV','DEC']
    return table.sort_values('YEAR')[names].to_numpy(float).ravel()
```

```python
t=pd.DataFrame({'YEAR':[2001,2000],**{m:[i+13,i+1] for i,m in enumerate(['JAN','FEB','MAR','APR','MAY','JUN','JUL','AUG','SEP','OCT','NOV','DEC'])}})
np.testing.assert_array_equal(calendar_vector(t),np.arange(1,25))
print("elnino.C1 checks passed")
```

**Why it works:** Column names are not alphabetically ordered times; the explicit month sequence preserves meaning.

## elnino.C2 — Fit only a supplied reference period

For finite values and corresponding month labels 1..12, return twelve monthly means; require all months.

Predict the check output before writing your function.

```python
def monthly_reference(values,months):
    a,m=np.asarray(values,float),np.asarray(months)
    if a.shape!=m.shape or set(m)!=set(range(1,13)) or not np.isfinite(a).all(): raise ValueError('All twelve months required')
    return np.array([a[m==j].mean() for j in range(1,13)])
```

```python
m=np.tile(np.arange(1,13),2)
np.testing.assert_allclose(monthly_reference(m.astype(float),m),np.arange(1,13))
print("elnino.C2 checks passed")
```

**Why it works:** Which rows you pass determines the fitted climatology. The function cannot infer the training boundary for you.

## elnino.C3 — Compute anomalies against a fixed baseline

Subtract each value’s month-specific mean from a supplied length-12 reference.

Predict the check output before writing your function.

```python
def monthly_anomalies(values,months,reference):
    a,m,r=np.asarray(values,float),np.asarray(months,int),np.asarray(reference,float)
    if a.shape!=m.shape or r.shape!=(12,) or not np.isin(m,np.arange(1,13)).all(): raise ValueError('Bad month or reference shape')
    return a-r[m-1]
```

```python
np.testing.assert_allclose(monthly_anomalies([25.5,22],[1,2],[24,20]+[0]*10),[1.5,2])
print("elnino.C3 checks passed")
```

**Why it works:** A positive anomaly is relative to a declared baseline, not an absolute category of climate behavior.
