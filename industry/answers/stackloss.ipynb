## stackloss.C1 — Compute MAE

Return mean absolute error for aligned nonempty finite vectors.

Predict the check output before writing your function.

```python
def mean_absolute_error_hand(actual,predicted):
    a,p=np.asarray(actual,float),np.asarray(predicted,float)
    if a.shape!=p.shape or a.ndim!=1 or not len(a) or not np.isfinite(a).all() or not np.isfinite(p).all(): raise ValueError('Finite aligned vectors required')
    return float(np.mean(np.abs(a-p)))
```

```python
assert mean_absolute_error_hand([10,12],[11,8])==2.5
assert mean_absolute_error_hand([1],[1])==0
print("stackloss.C1 checks passed")
```

**Why it works:** Absolute value removes the direction of error; averaging retains the response unit but can hide extreme failures.

## stackloss.C2 — Keep signed residuals

Return observed minus predicted; the sign convention must be explicit.

Predict the check output before writing your function.

```python
def signed_residual(actual,predicted):
    a,p=np.asarray(actual,float),np.asarray(predicted,float)
    if a.shape!=p.shape: raise ValueError('Shapes differ')
    return a-p
```

```python
np.testing.assert_array_equal(signed_residual([10,12],[11,8]),[-1,4])
np.testing.assert_array_equal(signed_residual([20],[15]),[5])
print("stackloss.C2 checks passed")
```

**Why it works:** A positive residual under this convention means the prediction was too low. Another convention is possible only when declared.

## stackloss.C3 — Leave one row out

Return all integer indices from 0 to n-1 except the specified omitted index.

Predict the check output before writing your function.

```python
def leave_out_indices(n,omitted):
    if n<2 or not 0<=omitted<n: raise ValueError('Invalid omission')
    return np.array([i for i in range(n) if i!=omitted],dtype=int)
```

```python
np.testing.assert_array_equal(leave_out_indices(4,1),[0,2,3])
assert 0 not in leave_out_indices(3,0)
print("stackloss.C3 checks passed")
```

**Why it works:** The scaler and model must both be refit on these indices, not only the model.
