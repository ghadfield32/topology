## modechoice.C1 — Recover one selected mode per person

Validate four modes and one selected alternative per individual, then return a Series of selected modes indexed by individual.

Predict the check output before writing your function.

```python
def selected_modes(table):
    result={}
    for person,g in table.groupby('individual',sort=True):
        if len(g)!=4 or set(g['mode'])!={1,2,3,4} or not g.choice.isin([0,1]).all() or g.choice.sum()!=1: raise ValueError('Invalid choice set')
        result[person]=int(g.loc[g.choice==1,'mode'].iloc[0])
    return pd.Series(result,dtype=int)
```

```python
t=pd.DataFrame({'individual':[1]*4,'mode':[1,2,3,4],'choice':[0,1,0,0]})
assert selected_modes(t).loc[1]==2
print("modechoice.C1 checks passed")
```

**Why it works:** Four alternative rows contain one decision. Invalid choice sets must not silently create extra labels.

## modechoice.C2 — Build a cheapest-mode baseline

Given an n-by-4 finite cost matrix ordered air/train/bus/car, return mode codes1..4; choose the first mode on ties.

Predict the check output before writing your function.

```python
def cheapest_mode(costs):
    a=np.asarray(costs,float)
    if a.ndim!=2 or a.shape[1]!=4 or not np.isfinite(a).all(): raise ValueError('Four finite offered costs required')
    return np.argmin(a,axis=1)+1
```

```python
np.testing.assert_array_equal(cheapest_mode([[4,2,3,1],[1,1,2,3]]),[4,1])
print("modechoice.C2 checks passed")
```

**Why it works:** A baseline embodies a simple assumption; it is not a claim that all people minimize ticket price.

## modechoice.C3 — Check disjoint travelers

Return True precisely when the two identifier collections share no member.

Predict the check output before writing your function.

```python
def groups_disjoint(train_ids,test_ids):
    return not bool(set(train_ids)&set(test_ids))
```

```python
assert groups_disjoint([1,1,2],[3,4])
assert not groups_disjoint([1,2],[2,3])
print("modechoice.C3 checks passed")
```

**Why it works:** Apply the same principle to athletes, patients and experiment runs. Disjoint rows alone do not establish disjoint entities.
