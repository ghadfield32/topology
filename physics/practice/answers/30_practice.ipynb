## 30.C1 — evidence_digest

Return the SHA-256 hash of an existing evidence file; do not invent a success for a missing file.

<details><summary>Hint</summary>Read bytes, not a filename string.</details>

```python
def evidence_digest(path):
    from hashlib import sha256
    from pathlib import Path
    return sha256(Path(path).read_bytes()).hexdigest()
```

## 30.C2 — heldout_mae

Compute mean absolute error for matching finite nonempty arrays.

<details><summary>Hint</summary>State the response unit separately; MAE inherits it.</details>

```python
def heldout_mae(prediction, reference):
    p=np.asarray(prediction,dtype=float);r=np.asarray(reference,dtype=float)
    if p.shape!=r.shape or p.size==0 or not np.isfinite(p).all() or not np.isfinite(r).all():raise ValueError('Matching finite nonempty arrays required.')
    return float(np.mean(np.abs(p-r)))
```

```python
assert heldout_mae([1,3],[2,3])==.5
with pytest.raises(ValueError):heldout_mae([],[])
print("30.C2: example checks passed; explain the assumptions.")
```

## 30.C3 — first_unfinished

Return the smallest integer stage not demonstrated or retained; return None if none.

<details><summary>Hint</summary>This reads declared statuses; it does not grade any answer.</details>

```python
def first_unfinished(stage_statuses):
    for key in sorted(stage_statuses,key=int):
     if stage_statuses[key] not in {'demonstrated','retained'}:return int(key)
    return None
```

```python
assert first_unfinished({'00':'retained','01':'practicing','02':'demonstrated'})==1
assert first_unfinished({'00':'demonstrated'}) is None
print("30.C3: example checks passed; explain the assumptions.")
```
