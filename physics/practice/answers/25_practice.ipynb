## 25.C1 — affine_correction

Project a point onto Az=b with full-row-rank A.

<details><summary>Hint</summary>Use solve(AAᵀ,Az-b), not an explicit inverse.</details>

```python
def affine_correction(z, A, b):
    z=np.asarray(z,dtype=float);A=np.asarray(A,dtype=float);b=np.asarray(b,dtype=float)
    if A.ndim!=2 or z.ndim!=1 or A.shape!=(len(b),len(z)) or np.linalg.matrix_rank(A)<len(b):raise ValueError('Full row rank and compatible dimensions required.')
    return z-A.T@np.linalg.solve(A@A.T,A@z-b)
```

```python
np.testing.assert_allclose(affine_correction([3,4],[[1,1]],[1]),[0,1])
with pytest.raises(ValueError):affine_correction([1,2],[[1,1],[2,2]],[1,2])
print("25.C1: example checks passed; explain the assumptions.")
```

## 25.C2 — circle_newton_step

Perform one linearized unit-circle constraint correction, not an exact radial projection.

<details><summary>Hint</summary>C=z·z-1 and A=2zᵀ.</details>

```python
def circle_newton_step(z):
    z=np.asarray(z,dtype=float);r=z@z
    if z.shape!=(2,) or not np.isfinite(r) or r==0:raise ValueError('Nonzero finite two-vector required.')
    return z-z*(r-1)/(2*r)
```

```python
np.testing.assert_allclose(circle_newton_step([2,0]),[1.25,0])
with pytest.raises(ValueError):circle_newton_step([0,0])
print("25.C2: example checks passed; explain the assumptions.")
```

## 25.C3 — tangent_component

Remove the radial component of v at unit q.

<details><summary>Hint</summary>Subtract q times the inner product.</details>

```python
def tangent_component(q, v):
    q=np.asarray(q,dtype=float);v=np.asarray(v,dtype=float)
    if q.shape!=v.shape or q.ndim!=1 or not np.isclose(q@q,1):raise ValueError('Unit point and matching vector required.')
    return v-q*(q@v)
```

```python
np.testing.assert_allclose(tangent_component([1,0],[2,3]),[0,3])
print("25.C3: example checks passed; explain the assumptions.")
```
