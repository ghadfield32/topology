## 21.C1 — centered_rate

Estimate f′(x) with the centered two-point difference; require finite positive h.

<details><summary>Hint</summary>Subtract symmetric values before dividing by 2h.</details>

```python
def centered_rate(f, x, h):
    if not np.isfinite(h) or h<=0: raise ValueError('Positive finite spacing required.')
    return (f(x+h)-f(x-h))/(2*h)
```

```python
assert abs(centered_rate(np.sin,.3,1e-4)-np.cos(.3))<1e-7
assert centered_rate(lambda x:x*x,2.,.1)==pytest.approx(4.)
with pytest.raises(ValueError): centered_rate(np.sin,0.,0.)
print("21.C1: example checks passed; explain the assumptions.")
```

## 21.C2 — spring_rhs

Return (q′,p′) for canonical state (q,p) and H=p²/(2m)+kq²/2; mass must be positive.

<details><summary>Hint</summary>Momentum is not velocity: q′=p/m.</details>

```python
def spring_rhs(z, mass, stiffness):
    z=np.asarray(z,dtype=float)
    if z.shape!=(2,) or not np.isfinite(z).all() or not np.isfinite(mass) or mass<=0 or not np.isfinite(stiffness) or stiffness<0: raise ValueError('Invalid spring state or parameters.')
    return np.array([z[1]/mass,-stiffness*z[0]])
```

```python
np.testing.assert_allclose(spring_rhs([2.,6.],2.,3.),[3.,-6.])
with pytest.raises(ValueError):spring_rhs([1,2],0,1)
print("21.C2: example checks passed; explain the assumptions.")
```

## 21.C3 — measured_secant

Calculate a measured response-per-input secant; reject coincident inputs.

<details><summary>Hint</summary>A secant is not necessarily a time derivative.</details>

```python
def measured_secant(x0, y0, x1, y1):
    a=np.asarray([x0,y0,x1,y1],dtype=float)
    if not np.isfinite(a).all() or x0==x1: raise ValueError('Finite observations with distinct inputs required.')
    return (y1-y0)/(x1-x0)
```

```python
assert measured_secant(24.41,.591,34.82,1.547)==pytest.approx(.956/10.41)
with pytest.raises(ValueError):measured_secant(1,2,1,3)
print("21.C3: example checks passed; explain the assumptions.")
```
