## 22.C1 — local_quadratic_flow

Evaluate x/(1-tx) only on the solution domain 1-tx>0.

<details><summary>Hint</summary>An algebraic value beyond blow-up is not a solution continuation.</details>

```python
def local_quadratic_flow(x, t):
    x=np.asarray(x,dtype=float)
    if not np.isfinite(x).all() or not np.isfinite(t) or np.any(1-t*x<=0):raise ValueError('Outside local flow domain.')
    return x/(1-t*x)
```

```python
x=np.array([-.3,.2]);np.testing.assert_allclose(local_quadratic_flow(local_quadratic_flow(x,.5),-.5),x)
with pytest.raises(ValueError):local_quadratic_flow(2.,.6)
print("22.C1: example checks passed; explain the assumptions.")
```

## 22.C2 — invert_coupling

Invert the stage coupling on an N×2 finite array.

<details><summary>Hint</summary>The first coordinate is unchanged, so solve the second scalar equation.</details>

```python
def invert_coupling(Y):
    Y=np.asarray(Y,dtype=float)
    if Y.ndim!=2 or Y.shape[1]!=2 or not np.isfinite(Y).all():raise ValueError('N by 2 finite input required.')
    x=Y[:,0]
    return np.column_stack([x,(Y[:,1]-.5*np.sin(x))*np.exp(-.2*np.tanh(x))])
```

```python
from shape_lab.physics.classical import coupling
X=np.array([[.2,1.],[-1.,2.]])
np.testing.assert_allclose(invert_coupling(coupling(X)[0]),X)
print("22.C2: example checks passed; explain the assumptions.")
```

## 22.C3 — coupling_log_density

For y=F(x), return log p_Y(y) from log p_X(x); x is first coordinate.

<details><summary>Hint</summary>Subtract the log absolute determinant.</details>

```python
def coupling_log_density(log_px, x):
    return np.asarray(log_px)-.2*np.tanh(np.asarray(x))
```

```python
np.testing.assert_allclose(coupling_log_density([-1.,-2.],[0.,1.]),[-1.,-2.-.2*np.tanh(1.)])
print("22.C3: example checks passed; explain the assumptions.")
```
