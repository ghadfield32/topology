## 29.C1 — balanced_pair_update

Add a message to i and its negative to j; leave the input unchanged.

<details><summary>Hint</summary>Copy the array, then use equal and opposite updates.</details>

```python
def balanced_pair_update(nodes, i, j, message):
    a=np.asarray(nodes,dtype=float).copy();v=np.asarray(message,dtype=float)
    if not 0<=i<len(a) or not 0<=j<len(a) or i==j or a.shape[1:]!=v.shape:raise ValueError('Valid endpoints and message shape required.')
    a[i]+=v;a[j]-=v
    return a
```

```python
x=np.zeros((3,2));out=balanced_pair_update(x,0,2,[1.,2.])
np.testing.assert_allclose(out.sum(axis=0),0);np.testing.assert_allclose(x,0)
print("29.C1: example checks passed; explain the assumptions.")
```

## 29.C2 — attention_invariants

Report constant preservation (rows) and total preservation (columns) for a square matrix.

<details><summary>Hint</summary>Test two different sums.</details>

```python
def attention_invariants(A):
    A=np.asarray(A,dtype=float)
    if A.ndim!=2 or A.shape[0]!=A.shape[1] or not np.isfinite(A).all():raise ValueError('Finite square matrix required.')
    return {'constant':bool(np.allclose(A.sum(axis=1),1)), 'total':bool(np.allclose(A.sum(axis=0),1))}
```

```python
assert attention_invariants([[1,0],[1,0]])=={'constant':True,'total':False}
assert attention_invariants(np.eye(2))=={'constant':True,'total':True}
print("29.C2: example checks passed; explain the assumptions.")
```

## 29.C3 — sensitivity_check

Estimate a scalar output sensitivity by a centered parameter perturbation.

<details><summary>Hint</summary>Compare this finite-difference program with autograd under matched solver settings.</details>

```python
def sensitivity_check(f, parameter, epsilon):
    if not np.isfinite(epsilon) or epsilon<=0:raise ValueError('Positive finite epsilon required.')
    return (f(parameter+epsilon)-f(parameter-epsilon))/(2*epsilon)
```

```python
assert sensitivity_check(lambda a:np.exp(-a),.3,1e-5)==pytest.approx(-np.exp(-.3),abs=1e-8)
print("29.C3: example checks passed; explain the assumptions.")
```
