## 23.C1 — canonical_momentum

Compute momentum for the stated constant-positive-mass kinetic model.

<details><summary>Hint</summary>Differentiate m*v²/2 with respect to v.</details>

```python
def canonical_momentum(mass, velocity):
    if not np.isfinite(mass) or mass<=0:raise ValueError('Positive mass required.')
    v=np.asarray(velocity,dtype=float)
    if not np.isfinite(v).all():raise ValueError('Finite velocity required.')
    return mass*v
```

```python
assert canonical_momentum(2.,3.)==6.
with pytest.raises(ValueError):canonical_momentum(0.,1.)
print("23.C1: example checks passed; explain the assumptions.")
```

## 23.C2 — hamiltonian_from_lagrangian

Return the quadratic spring H after the Legendre transform.

<details><summary>Hint</summary>Substitute v=p/m into p*v-L.</details>

```python
def hamiltonian_from_lagrangian(q, p, mass, stiffness):
    if not np.isfinite([q,p,mass,stiffness]).all() or mass<=0 or stiffness<0:raise ValueError('Invalid inputs.')
    return p*p/(2*mass)+stiffness*q*q/2
```

```python
assert hamiltonian_from_lagrangian(2.,6.,2.,3.)==15.
print("23.C2: example checks passed; explain the assumptions.")
```

## 23.C3 — regular_lagrangian_acceleration

Calculate scalar autonomous Euler–Lagrange acceleration; reject a near-zero velocity Hessian.

<details><summary>Hint</summary>Expand the time derivative of L_v using the chain rule.</details>

```python
def regular_lagrangian_acceleration(L_q, L_vq, L_vv, velocity):
    if not np.isfinite([L_q,L_vq,L_vv,velocity]).all() or abs(L_vv)<1e-12:raise ValueError('Regular finite scalar Lagrangian required.')
    return (L_q-L_vq*velocity)/L_vv
```

```python
assert regular_lagrangian_acceleration(-6,0,2,.4)==-3.
with pytest.raises(ValueError):regular_lagrangian_acceleration(1,0,0,2)
print("23.C3: example checks passed; explain the assumptions.")
```
