## 28.C1 — heat_multipliers

Return exp(-nu*(2*pi*k/length)²*time), for nonnegative diffusivity/time and positive length.

<details><summary>Hint</summary>The zero mode must have multiplier one.</details>

```python
def heat_multipliers(k, nu, time, length=1.):
    if not np.isfinite([nu,time,length]).all() or nu<0 or time<0 or length<=0:raise ValueError('Invalid heat parameters.')
    k=np.asarray(k,dtype=float)
    return np.exp(-nu*(2*np.pi*k/length)**2*time)
```

```python
m=heat_multipliers(np.arange(3),.03,.15)
assert m[0]==1 and m[2]<m[1]<m[0]
print("28.C1: example checks passed; explain the assumptions.")
```

## 28.C2 — branch_trunk_product

Combine branch[B,r] and trunk[N,r] by an unscaled inner product to output [B,N].

<details><summary>Hint</summary>This exercise deliberately omits the trained model's rank scaling and bias.</details>

```python
def branch_trunk_product(branch, trunk):
    b=np.asarray(branch);t=np.asarray(trunk)
    if b.ndim!=2 or t.ndim!=2 or b.shape[1]!=t.shape[1]:raise ValueError('Shared feature dimension required.')
    return b@t.T
```

```python
np.testing.assert_allclose(branch_trunk_product([[1,2]],[[3,4],[1,0]]),[[11,1]])
print("28.C2: example checks passed; explain the assumptions.")
```

## 28.C3 — disjoint_function_splits

Return True only when the three nonempty function-ID lists are unique and pairwise disjoint.

<details><summary>Hint</summary>Compare whole function IDs, not spatial row numbers.</details>

```python
def disjoint_function_splits(train_ids, validation_ids, test_ids):
    groups=[list(train_ids),list(validation_ids),list(test_ids)]
    sets=[set(x) for x in groups]
    return all(len(s)==len(g)>0 for s,g in zip(sets,groups)) and not any(sets[i]&sets[j] for i in range(3) for j in range(i))
```

```python
assert disjoint_function_splits([1,2],[3],[4])
assert not disjoint_function_splits([1,2],[2],[4])
assert not disjoint_function_splits([1,1],[2],[3])
print("28.C3: example checks passed; explain the assumptions.")
```
