## S06.C1 — Compute Git blob SHA1 including its type/length header.

Hint: Git object IDs are not plain SHA1(file bytes).

```python
def blob_hash(payload):
    import hashlib
    return hashlib.sha1(b"blob "+str(len(payload)).encode()+b"\0"+payload).hexdigest()
```

```python
assert blob_hash(b"hello\n")==sport.git_blob_sha1(b"hello\n")
print("S06.C1: check passed")
```

## S06.C2 — Identify Git LFS pointer text.

Hint: A pointer carries a hash and size, not the underlying tracking records.

```python
def is_lfs_pointer(payload):
    return payload.startswith(b"version https://git-lfs.github.com/spec/")
```
