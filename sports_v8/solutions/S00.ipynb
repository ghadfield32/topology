## S00.C1 — Convert feet to metres. Accept a numeric list and return a NumPy array.

Hint: A single factor changes units; it does not change coordinate origin.

```python
def feet_to_metres(values):
    return np.asarray(values, dtype=float) * 0.3048
```

```python
assert np.allclose(feet_to_metres([1,10]),[.3048,3.048])
print("S00.C1: check passed")
```

## S00.C2 — Count distinct athlete IDs, not rows.

## S00.C3 — Return elapsed seconds between two frame IDs; reject reversed frames and nonpositive fps.

Hint: The difference is in frame numbers, not row positions.

```python
def elapsed_seconds(start,end,fps):
    if end<=start or fps<=0:
        raise ValueError("Positive elapsed time and fps required")
    return (end-start)/fps
```

```python
assert elapsed_seconds(110,140,60)==.5
print("S00.C3: check passed")
```
