## S04.C1 — Find the latest available row index, or -1 before the first observation. Assume sorted times.

Hint: Right-sided search includes a sample at exactly the query time.

```python
def past_index(times,query):
    return int(np.searchsorted(np.asarray(times,float),query,side="right")-1)
```

```python
assert past_index([0,1,2],1.5)==1 and past_index([0,1],-.1)==-1
print("S04.C1: check passed")
```

## S04.C2 — Compute maximum displacement between corresponding points.

Hint: The correspondence of rows is an assumption; nearest-point matching is different.

```python
def max_displacement(x,y):
    return float(np.linalg.norm(np.asarray(x,float)-np.asarray(y,float),axis=1).max())
```

```python
assert np.isclose(max_displacement([[0,0],[1,1]],[[0,0],[1,1.2]]),.2)
print("S04.C2: check passed")
```

## S04.C3 — Return the matched-cloud edge-error bound for a nonnegative displacement bound.

Hint: Two endpoints can each move by delta.

```python
def edge_bound(delta):
    if delta<0:raise ValueError("Nonnegative displacement bound required")
    return 2*delta
```
