## S01.C1 — Return straight-line separation of two points.

Hint: Euclidean separation need not equal the unobserved travelled path.

```python
def separation(a,b):
    return float(np.linalg.norm(np.asarray(b,float)-np.asarray(a,float)))
```

```python
assert separation([0,0,0],[3,4,0])==5
print("S01.C1: check passed")
```

## S01.C2 — Return secant velocity for two positions and increasing times.

Hint: A secant is an interval average, not an exact instantaneous derivative.

```python
def secant(a,b,t0,t1):
    if t1<=t0:raise ValueError("Increasing times required")
    return (np.asarray(b,float)-np.asarray(a,float))/(t1-t0)
```

```python
assert np.allclose(secant([0,0,0],[.2,0,0],0,.1),[2,0,0])
print("S01.C2: check passed")
```

## S01.C3 — Compute cosine of the angle formed by two nonzero vectors.

Hint: Dot product over norm product; guard undefined zero rays.

```python
def angle_cosine(u,v):
    u,v=np.asarray(u,float),np.asarray(v,float)
    denom=np.linalg.norm(u)*np.linalg.norm(v)
    if denom==0:raise ValueError("Undefined angle")
    return float(np.clip(np.dot(u,v)/denom,-1,1))
```

```python
assert angle_cosine([1,0,0],[0,1,0])==0
print("S01.C3: check passed")
```
