## S02.C1 — Calculate height under constant downward acceleration.

Hint: Keep time relative to the stated origin.

```python
def vertical_height(z0,v0,g,t):
    t=np.asarray(t,float)
    return z0+v0*t-.5*g*t*t
```

```python
assert np.isclose(vertical_height(2,4,9.81,.2),2.6038)
print("S02.C1: check passed")
```

## S02.C2 — Calculate root-mean-square error of equal nonempty finite arrays.

Hint: Squaring, averaging and taking a square root retains output units.

```python
def rmse(actual,predicted):
    a,p=np.asarray(actual,float),np.asarray(predicted,float)
    if a.shape!=p.shape or not a.size or not np.isfinite(a).all() or not np.isfinite(p).all():raise ValueError("Invalid pair")
    return float(np.sqrt(np.mean((a-p)**2)))
```

```python
assert np.isclose(rmse([1,2],[1,4]),np.sqrt(2))
print("S02.C2: check passed")
```
